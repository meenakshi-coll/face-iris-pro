```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

```

```python
img = cv2.imread('sample.jpg')

if img is None:
    print(" Face image not loaded")
    exit()

img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Improve face detection
gray = cv2.equalizeHist(gray)

face_cascade = cv2.CascadeClassifier(
    cv2.data.haarcascades + 'haarcascade_frontalface_default.xml'
)

faces = face_cascade.detectMultiScale(
    gray,
    scaleFactor=1.1,
    minNeighbors=5
)

if len(faces) == 0:
    print(" No face detected")
    exit()

print(" Faces detected:", len(faces))

for (x, y, w, h) in faces:
    face = gray[y:y+h, x:x+w]

    # Resize for uniform features
    face_resized = cv2.resize(face, (64, 64))

    # Histogram Equalization
    face_resized = cv2.equalizeHist(face_resized)

    face_features = face_resized.flatten()

print("Face feature length:", len(face_features))
```

```python
iris = cv2.imread('iris.jpg', 0)

if iris is None:
    print(" Iris image not found")
    exit()

# Resize iris image
iris = cv2.resize(iris, (64, 64))

# Noise reduction
iris = cv2.GaussianBlur(iris, (5, 5), 0)

# Edge extraction
edges = cv2.Canny(iris, 50, 150)

iris_features = edges.flatten()

print("Iris feature length:", len(iris_features))
```
